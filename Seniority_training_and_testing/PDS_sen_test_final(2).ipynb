```python
from transformers import pipeline
import pandas as pd
import json
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

MODEL_PATH = "/content/drive/MyDrive/PDS_Snapaddy/models/bert_sen_Profs_JobBERT_PRO_uncapped"
JSON_PATH = "/content/drive/MyDrive/PDS_Snapaddy/Neuer Ordner/linkedin-cvs-annotated.json"

# 1. Modell laden (top_k=None weglassen für einfache Liste)
classifier = pipeline("text-classification", model=MODEL_PATH, tokenizer=MODEL_PATH)

# 2. Daten laden und flatten
job_df = pd.read_json(JSON_PATH)
records = [val for col in job_df.columns for val in job_df[col] if isinstance(val, dict)]
active = pd.DataFrame(records)[lambda df: df["status"] == "ACTIVE"].copy()
texts = active["position"].fillna("").tolist()

# 3. Vorhersagen (top_k=None im Call für alle Labels, aber top 1 reicht)
preds = classifier(texts, top_k=1)  # Einfache Liste von Dicts

active["predicted_seniority"] = [p[0]['label'] for p in preds]  # Sicher für nested
active["confidence"] = [p[0]['score'] for p in preds]

# 4. Label-Mapping laden
with open(f"{MODEL_PATH}/label_encoder.json", "r", encoding="utf-8") as f:
    le_info = json.load(f)
id2label = {int(k): v for k, v in le_info["mapping"].items()}

def decode_label(label_str):
    if isinstance(label_str, str) and label_str.startswith("LABEL_"):
        idx = int(label_str.split("_")[1])
        return id2label.get(idx, label_str)
    return label_str

active["predicted_seniority_decoded"] = active["predicted_seniority"].apply(decode_label)

# 5. Evaluation (nur mit GT)
df_eval = active.dropna(subset=["seniority", "predicted_seniority_decoded"])
y_true = df_eval["seniority"]
y_pred = df_eval["predicted_seniority_decoded"]

print("Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

# Confusion Matrix
labels = sorted(df_eval["seniority"].unique())
cm = confusion_matrix(y_true, y_pred, labels=labels)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", xticklabels=labels, yticklabels=labels, cmap="Blues")
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Seniority Confusion Matrix")
plt.tight_layout(); plt.show()

# Fehler speichern
errors = df_eval[df_eval["seniority"] != df_eval["predicted_seniority_decoded"]]
active.to_csv("/content/drive/MyDrive/PDS_Snapaddy/linkedin_seniority_predictions.csv", index=False)
df_eval.to_csv("/content/drive/MyDrive/PDS_Snapaddy/linkedin_seniority_eval.csv", index=False)
errors.to_csv("/content/drive/MyDrive/PDS_Snapaddy/linkedin_seniority_errors.csv", index=False)
```

```python
from transformers import pipeline
import pandas as pd
import json, requests
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import googletrans  # pip install googletrans==4.0.0-rc1

MODEL_PATH = "/content/drive/MyDrive/PDS_Snapaddy/models/bert_sen_Profs_JobBERT_PRO_uncapped"
JSON_PATH = "/content/drive/MyDrive/PDS_Snapaddy/Neuer Ordner/linkedin-cvs-annotated.json"

# 1. Translator + Classifier laden
translator = googletrans.Translator()  # Free, schnell!
classifier = pipeline("text-classification", model=MODEL_PATH, tokenizer=MODEL_PATH)

def translate_to_en(text):
    """Übersetzt zu EN (batch-safe)"""
    if pd.isna(text) or not text.strip():
        return ""
    try:
        result = translator.translate(text.strip(), dest='en')
        return result.text
    except:
        return text.strip()  # Fallback Original

# 2. Daten laden
job_df = pd.read_json(JSON_PATH)
records = [val for col in job_df.columns for val in job_df[col] if isinstance(val, dict)]
active = pd.DataFrame(records)[lambda df: df["status"] == "ACTIVE"].copy()

# 3. **Übersetzen** (langsamer, aber once!)
print("🔄 Übersetze", len(active), "Texts...")
active["position_en"] = active["position"].fillna("").apply(translate_to_en)
texts_en = active["position_en"].fillna("").tolist()

# 4. Predictions auf EN
preds = classifier(texts_en, top_k=1)
active["predicted_seniority"] = [p[0]['label'] for p in preds]
active["confidence"] = [p[0]['score'] for p in preds]

# 5. Decode (rest gleich)
with open(f"{MODEL_PATH}/label_encoder.json", "r", encoding="utf-8") as f:
    le_info = json.load(f)
id2label = {int(k): v for k, v in le_info["mapping"].items()}

def decode_label(label_str):
    if isinstance(label_str, str) and label_str.startswith("LABEL_"):
        idx = int(label_str.split("_")[1])
        return id2label.get(idx, label_str)
    return label_str

active["predicted_seniority_decoded"] = active["predicted_seniority"].apply(decode_label)

# 6. Eval (rest identisch)
df_eval = active.dropna(subset=["seniority", "predicted_seniority_decoded"])
y_true = df_eval["seniority"]
y_pred = df_eval["predicted_seniority_decoded"]

print("✅ EN-Übersetzt! Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

# Confusion Matrix
labels = sorted(df_eval["seniority"].unique())
cm = confusion_matrix(y_true, y_pred, labels=labels)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", xticklabels=labels, yticklabels=labels, cmap="Blues")
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Seniority (EN-translated)")
plt.tight_layout(); plt.show()

# Save mit EN-Texts
active.to_csv("/content/drive/MyDrive/PDS_Snapaddy/linkedin_seniority_en_translated.csv", index=False)
df_eval.to_csv("/content/drive/MyDrive/PDS_Snapaddy/linkedin_seniority_eval_en.csv", index=False)
```
